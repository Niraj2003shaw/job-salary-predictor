FEATURE ENGINEERING

```python
model_df = salary_df.copy()

print("Model dataset:", model_df.shape)
```

```python
salary_df["target_salary"] = (
    salary_df["minimumSalary"] + salary_df["maximumSalary"]
) / 2
```

```python
model_df["target_salary"] = (
    model_df["minimumSalary"] + model_df["maximumSalary"]
) / 2
```

```python
model_df["log_target_salary"] = np.log1p(
    model_df["target_salary"]
)
```

```python
model_df[
    ["salary", "minimumSalary", "maximumSalary", "target_salary"]
].head(10)
```

```python
model_df["log_target_salary"] = np.log1p(
    model_df["target_salary"]
)
```

```python
model_df[
    ["target_salary", "log_target_salary"]
].head(10)
```

NORMALIZE LOCATION

```python
major_cities = [
    "Bengaluru",
    "Mumbai",
    "Delhi",
    "Gurugram",
    "Noida",
    "Hyderabad",
    "Chennai",
    "Pune",
    "Kolkata",
    "Ahmedabad",
    "Jaipur",
    "Surat",
    "Vadodara",
    "Kochi",
    "Coimbatore",
    "Nagpur",
    "Thane",
    "Navi Mumbai",
    "Indore",
    "Lucknow",
    "Chandigarh",
    "Bhopal",
    "Patna",
    "Bhubaneswar",
    "Ranchi"
]

def normalize_location(location):
    location = str(location).strip()

    if "remote" in location.lower():
        return "Remote"

    for city in major_cities:
        if city.lower() in location.lower():
            return city

    return "Other"

model_df["location_clean"] = model_df["location"].apply(
    normalize_location
)
```

```python
model_df["title_clean"] = (
    model_df["title"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.strip()
)
```

```python
model_df[["title", "title_clean"]].head(15)
```

```python
model_df["title_clean"] = (
    model_df["title_clean"]
    .str.replace(r"[^a-zA-Z0-9\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
model_df[["title", "title_clean"]].head(20)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
title_vectorizer = TfidfVectorizer(
    max_features=3000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)
```

```python
title_features = title_vectorizer.fit_transform(
    model_df["title_clean"]
)

print("Title feature matrix shape:", title_features.shape)
```

```python
print(
    title_vectorizer.get_feature_names_out()[:100]
)
```

```python
print("Total title features:", len(
    title_vectorizer.get_feature_names_out()
))
```

```python
model_df["skills_clean"] = (
    model_df["tagsAndSkills"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace(",", " ", regex=False)
    .str.replace(r"[^a-zA-Z0-9\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
model_df[
    ["tagsAndSkills", "skills_clean"]
].head(10)
```

```python
skill_vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)
```

```python
skill_features = skill_vectorizer.fit_transform(
    model_df["skills_clean"]
)
```

```python
print("Skill feature matrix shape:", skill_features.shape)
```

```python
print(
    "Total skill features:",
    len(skill_vectorizer.get_feature_names_out())
)
```

```python
print("Model rows:", len(model_df))
print("Title TF-IDF rows:", title_features.shape[0])
print("Skill TF-IDF rows:", skill_features.shape[0])
```
